```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("/content/00 customer.csv")
df.sample(5)
```

```python
print(df['review'].unique())
print(df['education'].unique())
print(df['purchased'].unique())
```

```python
df = df.iloc[:,2:]
df.sample(5)
```

```python
x = df[['review', 'education']]
y = df['purchased']

print(x.head())
print()
print()
print()
print(y.head())
```

```python
from sklearn.preprocessing import OrdinalEncoder

oe = OrdinalEncoder(categories = [['Poor','Average','Good'],['School','UG','PG']]).set_output(transform='pandas')
```

```python
rev_edu = oe.fit_transform(x)
rev_edu
```

```python
oe.categories_
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
purchased = le.fit_transform(y)

y_df = pd.DataFrame(purchased, columns=['Purchased'])
y_df
```
